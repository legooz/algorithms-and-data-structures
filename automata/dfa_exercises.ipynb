```python
START = "q0"
octet = ""  

def is_non_ws(c):
    return not c.isspace()

def is_ws(c):
    return c.isspace()

def is_digit(c):
    return c.isdigit()

def is_alpha(c):
    return c.isalpha()




def step(state, ch, *, start=START):
    global octet 

    match state:
        case "q0":
            if is_ws(ch):
                return "q1"
            return start

        case "q1":
            if is_ws(ch):
                return "q1"
            if ch == "h":
                return "a0"
            if ch == "0" or ch == "1" or ch == "2" or ch == "3" or ch == "4" or ch == "5" or ch == "6" or ch == "7" or ch == "8" or ch == "9":
                octet = ch 
                return "c0"
            # if ch == "1" or ch == "0":  return "c0"
            # if ch == "2":   return "d0"
            # if ch == "3" or ch == "4" or ch == "5" or ch == "6" or ch == "7" or ch == "8" or ch == "9": return "e0"
            if is_non_ws(ch) and ch != "h" and ch != "@":
                return "b0"
            return start

        case "a0":
            if ch == "t":
                return "a1"
            if is_non_ws(ch) and ch != "t" and ch != "@":
                return "b0"
            return start

        case "a1":
            if ch == "t":
                return "a2"
            if is_non_ws(ch) and ch != "t" and ch != "@":
                return "b0"
            return start

        case "a2":
            if ch == "p":
                return "a3"
            if is_non_ws(ch) and ch != "p" and ch != "@":
                return "b0"
            return start

        case "a3":
            if is_non_ws(ch):
                return "a3"
            if is_ws(ch):
                return "a4"
            return start

        case "a4":  # accepting state
            if is_ws(ch):
                return "a4"
            return start

        case "b0":
            if ch == ".":
                return "b1"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return start

        case "b1":
            if ch == "c":
                return "z0"
            if ch == "n":
                return "y0"
            if ch == "g":
                return "x0"
            if ch == "e":
                return "w0"
            if ch == "o":
                return "v0"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return start

        case "z0":
            if ch == "o":
                return "z1"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "z1":
            if ch == "m":
                return "b2"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "y0":
            if ch == "e":
                return "y1"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "y1":
            if ch == "t":
                return "b2"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "x0":
            if ch == "o":
                return "x1"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "x1":
            if ch == "v":
                return "b2"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "w0":
            if ch == "d":
                return "w1"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "w1":
            if ch == "u":
                return "b2"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "v0":
            if ch == "r":
                return "v1"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "v1":
            if ch == "g":
                return "b2"
            if is_non_ws(ch) and ch != "@":
                return "b0"
            return state

        case "b2":
            if is_non_ws(ch):
                return "b2"
            if is_ws(ch):
                return "b3"
            return start

        case "b3":
            return start

        # URL with IP address

        case "c0":
            if is_digit(ch):
                octet += ch              
                return "c1"
            if ch == ".":
                octet = ""
                return "c3"

            return start
    
        case "c1":
            if ch == ".":
                octet = ""
                return "c3"

            if is_digit(ch):
                octet += ch 
                if octet != "" and int(octet) <= 255: 
                    #print(f"octet so far: {int(octet)}")
                    return "c2"

            octet = ""
            return start
        
        case "c2":
            if ch == ".": return "c3"
            return start
        
        case "c3":
            if ch == "0" or ch == "1" or ch == "2" or ch == "3" or ch == "4" or ch == "5" or ch == "6" or ch == "7" or ch == "8" or ch == "9":
                octet = ch 
                return "c4"
            return start

        case "c4":
            if is_digit(ch):
                octet += ch
                return "c5"
            if ch == ".":
                octet = ""
                return "c7"
            return start

        case "c5":
            if ch == ".":
                octet = ""
                return "c7"
            if is_digit(ch):
                octet += ch
                if octet != "" and int(octet) <= 255:
                    return "c6"
            octet = ""
            return start

        case "c6":
            if ch == ".":
                return "c7"
            return start

        case "c7":
            if ch == "0" or ch == "1" or ch == "2" or ch == "3" or ch == "4" or ch == "5" or ch == "6" or ch == "7" or ch == "8" or ch == "9":
                octet = ch
                return "c8"
            return start

        case "c8":
            if is_digit(ch):
                octet += ch
                return "c9"
            if ch == ".":
                octet = ""
                return "c11"
            return start

        case "c9":
            if ch == ".":
                octet = ""
                return "c11"
            if is_digit(ch):
                octet += ch
                if octet != "" and int(octet) <= 255:
                    return "c10"
            octet = ""
            return start

        case "c10":
            if ch == ".":
                return "c11"
            return start

        case "c11":
            if ch == "0" or ch == "1" or ch == "2" or ch == "3" or ch == "4" or ch == "5" or ch == "6" or ch == "7" or ch == "8" or ch == "9":
                octet = ch
                return "c12"
            return start

        case "c12":
            if is_digit(ch):
                octet += ch
                return "c13"
            if is_non_ws(ch):
                octet = ""
                return "c14"
            if is_ws(ch):
                octet = ""
                return "c15"
            return start

        case "c13":
            if is_digit(ch):
                if len(octet) < 3:
                    octet += ch
                    return "c13"
                octet = ""
                return start
            if is_non_ws(ch):
                octet = ""
                return "c14"
            if is_ws(ch):
                octet = ""
                return "c15"
            return start

        case "c14":
            if is_non_ws(ch):
                return "c14"
            if is_ws(ch):
                return "c15"
            return start

        case "c15":
            return start

        
        case _:
            return start


def run_dfa(start, accepts, s):
    state = start
    for ch in s:
        state = step(state, ch, start=start)
    return state in accepts


accepts = {"a4", "b3", "c15", "d15"}

text = (
    "Hello Myra, you can email me at automata907@alaska.org or visit the website at "
    "alaska.org/automata and learn about the theory of computing! Or call me at 907.786.1234. "
    "Another way to learn about automata is to visit 192.229.210.176/automata_theory/index.htm or "
    "to go to https://en.wikipedia.org/wiki/Automata_theory for more information. "
    "Mr.Foo posted in section 999.10.34.997 about the equivalence between finite state automata "
    "and regular expressions. He regularly updates www.automata.com with the latest developments."
)

print("URL(s) found:\n")
for token in text.split():
    padded = " " + token + " "
    if run_dfa(START, accepts, padded):
        print(token)
```
