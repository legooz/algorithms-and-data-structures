```python
def computeHash(s):
  hash = 0
  for ch in s:
    hash = hash + ord(ch)    # ord gives the ASCII value of character ch
  return hash % 10           # Gives number 0 to 9;  a prime number is better
```

```python
hash_value = computeHash("dog")
print("The hash value for 'dog' is:", hash_value)
```

```python
class HashTable:
  def __init__(self, size = 10):
    # Initialize list of length "size" so we use it like an array
    self.array = [None] * size

  def computeHash(self, s):
    hash = 0
    for ch in s:
      hash = hash + ord(ch)    # ord gives the ASCII value of character ch
    return hash % len(self.array)

  def add(self, key, payload):
    hash = self.computeHash(key)
    if self.array[hash] is not None:
      # If key is found in the list then update its value
      for pair in self.array[hash]:
        if (pair[0]==key):
          pair[1] = payload
          return
      # If we get here then we didn't find the key, so add it
      self.array[hash].append([key, payload])
    else:
      # The hash index is empty, create first entry
      self.array[hash] = [[key, payload]]

  def retrieve(self, key):
    hash = self.computeHash(key)
    if self.array[hash] is None:
      return None
    else:
      for pair in self.array[hash]:
        if (pair[0]==key):
          return(pair[1])
    return None   # Key does not exist or index empty
```

```python
# Demo code
h = HashTable(10)
h.add("dog", "This is a dog and could be any object")
h.add("cat", "This is a cat")
h.add("turtle", "This is a turtle")
h.add("bird", "This is a bird")
h.add("Wolverine", "This is a fuzzy animal")

print("dog?",h.retrieve("dog"))
print("cat?",h.retrieve("cat"))
print("fish?",h.retrieve("fish"))
print("bird?",h.retrieve("bird"))
print("Wolverine?", h.retrieve("Wolverine"))
print(h.array)
```
