```python
class Heap:
    def __init__(self):
        self.data = []   # Empty heap

    def extractMax(self):
        if len(self.data) == 0:
            return None  # Handle case of empty heap
        max_element = self.data[0]
        self.data[0] = self.data[-1]
        self.data.pop()   # removes last element
        self.heapify(0)
        return max_element

    def heapify(self, i):
        largest = i    # Initialize largest as root
        left = 2*i + 1   # Formula for 0-based array
        right = 2*i + 2
        heapSize = len(self.data)
        if (left < heapSize and self.data[left] > self.data[largest]):
            largest = left
        if (right < heapSize and self.data[right] > self.data[largest]):
            largest = right

        if (largest != i):
            self.data[i], self.data[largest] = self.data[largest], self.data[i]  # Swap
            self.heapify(largest)

    def buildHeap(self, sourceList):
        self.data = copy.deepcopy(sourceList)   # Makes deep copy of list
        for i in range((len(self.data) - 2) // 2, -1, -1):  # Start from last non-leaf node
            self.heapify(i)

    def insert(self, element):

        self.data.append(element)  # Add the new element to the end of the list
        current = len(self.data) - 1  # Index of the newly added element
        # Bubble up (or sift up) the new element as necessary to maintain heap property
        while current > 0:
            parent = (current - 1) // 2
            if self.data[current] > self.data[parent]:  # For max heap
                self.data[current] = self.data[parent]
                self.data[parent] = element
                current = parent
            else:
                break
```

```python
import copy
myheap = Heap()
predefined_list = [50, 45, 15, 5, 4, 3, 10, 2]
myheap.buildHeap(predefined_list)

# Insert the new element
new_element = 68
myheap.insert(new_element)

# Display the modified heap
myheap.data
```
