# Linked List Operations

Insert nodes, find values and remove the second node from a singly linked list.

[Attribution and source provenance](../NOTICE.md)


```python
class IntNode:
    def __init__(self, info=None, next=None):
        self.info = info
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, item):
        node = IntNode(item, self.head)
        self.head = node

    def find(self, item):
        temp = self.head
        while temp is not None:
            if temp.info == item:
                return temp
            temp = temp.next
        return None

    def deleteSecond(self):
        if self.head and self.head.next:
            temp = self.head.next
            self.head.next = self.head.next.next
            del temp

    def print(self):
        temp = self.head
        print()
        while temp:
            print(temp.info)
            temp = temp.next
        print()


if __name__ == "__main__":
    list = LinkedList()
    list.insert(10)
    list.insert(20)
    list.insert(30)

    # Test the find function
    nptr = list.find(20)
    if nptr is not None:
        print("Found", nptr.info)
    else:
        print("Didn't find 20")
    nptr = list.find(40)
    if nptr is not None:
        print("Found", nptr.info)
    else:
        print("Didn't find 40")


    list.print()


    list.deleteSecond()

    list.print()
```
