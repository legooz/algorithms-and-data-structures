# Movie Rating Binary Search Tree

Insert movie ratings, traverse them in order and remove nodes with zero, one or two children. The original demo renders a Graphviz file and opens a viewer.

[Attribution and source provenance](../NOTICE.md)


```python
from graphviz import Digraph

class BSTNode:
    def __init__(self, movie):
        self.movie = movie
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, movie):
        if self.root is None:
            self.root = BSTNode(movie)
        else:
            self._insert_recursive(self.root, movie)

    def _insert_recursive(self, current_node, movie):
        if movie['rating'] < current_node.movie['rating']:
            if current_node.left is None:
                current_node.left = BSTNode(movie)
            else:
                self._insert_recursive(current_node.left, movie)
        else:
            if current_node.right is None:
                current_node.right = BSTNode(movie)
            else:
                self._insert_recursive(current_node.right, movie)

    def inorder_traversal(self):
        movies = []
        self._inorder_traversal_recursive(self.root, movies)
        return movies

    def _inorder_traversal_recursive(self, current_node, movies):
        if current_node:
            self._inorder_traversal_recursive(current_node.left, movies)
            movies.append(current_node.movie)
            self._inorder_traversal_recursive(current_node.right, movies)

    def visualize(self):
        if self.root is None:
            print("The tree is empty.")
            return

        dot = Digraph(comment='BST')
        self._visualize_recursive(self.root, dot)
        return dot

    def _visualize_recursive(self, node, dot, parent=None, edge_label=''):
        if node is not None:
            node_id = str(hash(node))
            dot.node(node_id, f"{node.movie['title']} ({node.movie['rating']})")
            if parent is not None:
                dot.edge(str(hash(parent)), node_id, label=edge_label)

            self._visualize_recursive(node.left, dot, node, 'L')
            self._visualize_recursive(node.right, dot, node, 'R')

    def delete(self, rating):
        self.root = self._delete_recursive(self.root, rating)

    def _delete_recursive(self, current_node, rating):
        if current_node is None:
            return current_node

        if rating < current_node.movie['rating']:
            current_node.left = self._delete_recursive(current_node.left, rating)
        elif rating > current_node.movie['rating']:
            current_node.right = self._delete_recursive(current_node.right, rating)
        else:
            if current_node.left is None:
                return current_node.right
            elif current_node.right is None:
                return current_node.left

            temp = self._min_value_node(current_node.right)
            current_node.movie = temp.movie
            current_node.right = self._delete_recursive(current_node.right, temp.movie['rating'])
        return current_node

    def _min_value_node(self, node):
        current = node
        while current.left is not None:
            current = current.left
        return current
```

```python
bst = BinarySearchTree()

# Insert sample movies into the BST
bst.insert({"title": "Inception", "rating": 8.8})
bst.insert({"title": "The Matrix", "rating": 8.6})
bst.insert({"title": "Interstellar", "rating": 8.4})
bst.insert({"title": "The Prestige", "rating": 8.1})
bst.insert({"title": "The Lion King", "rating": 8.3})
bst.insert({"title": "Fight Club", "rating": 8.5})
bst.insert({"title": "Forrest Gump", "rating": 8.7})
bst.insert({"title": "The Dark Knight", "rating": 9.0})
bst.insert({"title": "Pulp Fiction", "rating": 8.9})
bst.insert({"title": "Shawshank Redemption", "rating": 9.3})

bst.delete(8.3)
bst.delete(8.8)

#visualization
dot = bst.visualize()
dot.render('bst_tree', view=True)
```
