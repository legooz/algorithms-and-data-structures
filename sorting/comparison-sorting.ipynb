```python
def bubbleSort(alist):
    iterations = 0  # Counter for the number of iterations
    for passnum in range(len(alist)-1, 0, -1):
        swapped = False
        for i in range(passnum):
            iterations += 1  # Increment the counter for each comparison
            if alist[i] > alist[i+1]:
                alist[i], alist[i+1] = alist[i+1], alist[i]
    return iterations  # Return the total number of iterations

# Function to demonstrate bubble sort with different array sizes
def demonstrate_bubble_sort():
    sizes = [5, 10, 15, 20, 25]  # Different sizes of arrays
    for size in sizes:
        data = list(range(size, 0, -1))
        print(f"Array size: {size}")
        print("Before sorting:")
        print(data)
        iterations = bubbleSort(data)
        print("After sorting:")
        print(data)
        print(f"Total iterations for size {size}: {iterations}\n")

demonstrate_bubble_sort()
```

```python
import matplotlib.pyplot as plt


# Function to run bubble sort and collect data on iterations
def collect_iterations(sizes):
    iterations_counts = []
    for size in sizes:
        data = list(range(size, 0, -1))
        iterations = bubbleSort(data)
        iterations_counts.append(iterations)
    return iterations_counts

sizes = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]  # Array sizes
iterations_counts = collect_iterations(sizes)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(sizes, iterations_counts, 'o-', label='Bubble Sort Iterations')
plt.plot(sizes, [x**2 for x in sizes], '--', label='$n^2$ for comparison')
plt.title('Bubble Sort Complexity Demonstration')
plt.xlabel('Array Size (n)')
plt.ylabel('Iterations (Number of Operations)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
def bubbleSort(alist):
    for passnum in range(len(alist)-1, 0, -1):
        swapped = False
        for i in range(passnum):
            if alist[i] > alist[i+1]:
                # Swap alist[i] and alist[i+1]
                alist[i], alist[i+1] = alist[i+1], alist[i]
                swapped = True
        print(alist)
        if not swapped:  # If no swap has occurred in this pass, break out of the loop
            break


data = [10, 20, 30, 40, 50, 60, 70, 90, 80, 100]
#print("Before sorting:")
#print(data)
bubbleSort(data)
#print("After sorting:")
#print(data)
```

```python
def selectionSort(alist):
    for fillslot in range(len(alist)-1):   # count 0 to len-2
        positionOfMin=fillslot
        for i in range(fillslot,len(alist)):  # count fillslot to len-1
            if alist[i]<alist[positionOfMin]:
                positionOfMin = i
        # swap alist[fillslot], alist[positionOfMin]
        temp = alist[fillslot]
        alist[fillslot] = alist[positionOfMin]
        alist[positionOfMin] = temp
        print(alist)
```

```python
def selectionSort(alist):
    iterations = 0  # Counter for the number of iterations
    for fillslot in range(len(alist)-1):   # count 0 to len-2
        positionOfMin = fillslot
        for i in range(fillslot+1, len(alist)):  # count fillslot+1 to len-1
            iterations += 1  # Increment the counter for each comparison
            if alist[i] < alist[positionOfMin]:
                positionOfMin = i
        # Swap alist[fillslot], alist[positionOfMin] using Python's tuple unpacking
        alist[fillslot], alist[positionOfMin] = alist[positionOfMin], alist[fillslot]
        print(alist)  # Print the list after each swap
    print(f"Total iterations: {iterations}")
    return iterations

# Function to demonstrate selection sort with different array sizes
def demonstrate_selection_sort():
    sizes = [5, 10, 15, 20, 25]  # Different sizes of arrays
    iterations_counts = []
    for size in sizes:
        data = list(range(size, 0, -1))  # Generate a reversed list of the given size
        print(f"\nArray size: {size}")
        iterations = selectionSort(data)
        iterations_counts.append(iterations)

    return sizes, iterations_counts

sizes, iterations_counts = demonstrate_selection_sort()

# Now we plot
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(sizes, iterations_counts, marker='o', linestyle='-', color='b')
plt.plot(sizes, [x**2 for x in sizes], linestyle='--', color='r', label='$n^2$ reference')
plt.title('Selection Sort Iterations vs. Array Size')
plt.xlabel('Array Size (n)')
plt.ylabel('Iterations')
plt.legend(['Selection Sort Iterations', '$n^2$ reference'])
plt.grid(True)
plt.show()
```
