# Insertion Sort

Sort an integer list in place by inserting each element into the ordered prefix.

[Attribution and source provenance](../NOTICE.md)


```python
# Insrtion sort in Python
def insertion_sort(arr):
    for j in range(1, len(arr)):
        key = arr[j]
        i = j - 1
        while i >= 0 and arr[i] > key:
            arr[i + 1] = arr[i]
            i -= 1
        arr[i + 1] = key


arr = [5, 3, 8, 2, 10, 7]
insertion_sort(arr)
print(arr)
```
